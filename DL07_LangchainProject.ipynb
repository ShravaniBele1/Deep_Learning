```python

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnablePassthrough
from langchain_groq import ChatGroq

def load_llm():
    return ChatGroq(
        model="llama-3.1-8b-instant",
        temperature=0.7
    )

def build_chain(llm):

    prompt = ChatPromptTemplate.from_template(
        """
You are a creative social media assistant.

Generate a catchy caption for the given input.

Make it:
- Short
- Engaging
- Creative
- Add emojis if suitable

Input:
{input}

Caption:
"""
    )

    chain = (
        {"input": RunnablePassthrough()}
        | prompt
        | llm
    )

    return chain

def main():

    print("\n===== AI CAPTION GENERATOR READY =====\n")

    llm = load_llm()
    chain = build_chain(llm)

    while True:
        user_input = input("Enter topic/text (or 'exit'): ")

        if user_input.lower() == "exit":
            break

        response = chain.invoke(user_input)

        print("\nGenerated Caption:\n", response.content)
        print("\n" + "="*50)

if __name__ == "__main__":
    main()
```
